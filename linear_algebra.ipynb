#### 1. Dot product (Inner Produc)

For Two vectors a = [a1, a2, a3,......., an] and b = [b1, b2, b3, ....., bn]
     
$$
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^n a_i b_i
$$

it combines two vectors into single scalar

Geomentry meaning

$$
\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\|\|\mathbf{b}\|\cos(\theta)
$$
where θ is the angle between them.
* if result > 0 --> acute angle (similar direction)
* if result < 0 --> obtuse angle (opposite direction)
* if result = 0 --> perpendicular (independent).

ML Example 
* in word embeddings (Word2Vec, Glove), dot product measures semantic relatedness.
 * e.g., "king" . "queen" > "king" . "car".

* Attention in Transformers: Query and key vectors use dot products to measure relevance.

* Neural networks: Each neuron computes a weighted sum (dot product between inputs and weights)

````python
import numpy as np 
import math

# Define two vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# ___ Algebraic defination __
dot_algebraic = np.dot(a, b)  # sum(ai * bi)

# ---Geometric meaning ---
norm_a = np.linalg.norm(a)  # ||a||
norm_b = np.linalg.norm(b)  # ||b||
cos_theta = dot_algebraic/(norm_a * norm_b) # cos(theta)
theta = math.degrees(math.acos(cos_theta))  # angle in degrees
dot_geometric = norm_a * norm_b * cos_theta


print("Dot product (algebraic):", dot_algebraic)
print("Dot product (geomtric):", dot_geometric)
print("Norm of a:", norm_a)
print("Norm of b:", norm_b)
print("cos(theta):", cos_theta)
print("Angle between vectors (degrees):", theta)
````

 $$
\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i
$$

##### Example:

````python
import numpy as np

a = np.array([1,2])
b = np.array([3,4])

dot_product = np.dot(a, b)
dot_product
````

### 3. Norm (Length of a vector)

The norm is the length of the vector.
it's like asking: "How long is the arrow?"

#### Formula (Euclidean norm):

$$
\|\mathbf{a}\| = \sqrt{\sum_{i=1}^{n} a_i^2}
$$            

Example:
 
$$
\|[3,4]\| = \sqrt{3^2 + 4^2} = 5
$$

````python
a = np.array([3, 4])

norm_a = np.linalg.norm(a)
norm_a
````

#### 4. Cosine similarity

Cosine similarity asks:
   "How similar are two vectors in terms of direction?"

* If vectors point the same way cosine similarity = 1
* If vectors are perpendicular (90 degrees), similarity = 0
* If vectors point in opposite directions, similarity = -1

##### Formula:
$$
\text{cosine\_similarity}(\mathbf{a}, \mathbf{b}) =
\frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\| \|\mathbf{b}\|}
$$

It's the dot product divided by the product of lengths

````python
a = np.array([1,2])
b = np.array([3,4])

cos_sim = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
cos_sim
````

#### 5 . Manhattan Norm (L1 Norm)

##### Idea
The Manhattan norm measures **distance along grid lines**, like a taxi driving in a city.  
It adds up the absolute values of all components.

##### Formula
For a vector $\mathbf{a} = [a_1, a_2, ..., a_n]$:

$$
\|\mathbf{a}\|_1 = |a_1| + |a_2| + \dots + |a_n|
$$

##### Example
For $\mathbf{a} = [3, -4, 1]$:

$$
\|\mathbf{a}\|_1 = |3| + |-4| + |1| = 3 + 4 + 1 = 8
$$

##### Python Code
```python

````python
import numpy as np

a = np.array([3, -4, 1])
manhattan_norm = np.sum(np.abs(a))
manhattan_norm
````

## Matrices & Matrix Multiplication in ML

#### 1. What is a Matrix ?

A matrix is just a grid of numbers ---- like a spreadsheet:

$$
A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix} \quad
$$

* Rows --> horizontal 
* Columns --> Vertical

In ML :
* Data: each row is a data sample, each column is a feature.
* Weights in neural networks are stored as matrices.

#### 2. Matrix Multiplication 

* Matrix multiplication is like a super-powered dot product 
* Each element of the new matix is the dot product od a row from the first matrix and a column from the second.

##### Formula:
 If A is m x n and B is n x p:

$$
C_{i,j} = \sum_{k=1}^{n} A_{i,k} \cdot B_{k,j}
$$

##### Example:

$$
A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}, \quad
B = \begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}, \quad
A \times B = 
\begin{bmatrix} 19 & 22 \\ 43 & 50 \end{bmatrix}
$$



````python
import numpy as np

A = np.array([[1,2],
             [3,4]])

B = np.array([[5, 6],
              [7, 8]])

#C = A X B = 
AXB  = [[((1 * 5) + (2 * 7)), ((1 * 6) + (2 * 8))], 
      [[(3 * 5)+(4 * 7), ((3 * 6)+(4*8))]]]
print(AXB)

C = np.dot(A, B)
C
````

#### 3. Why ML Loves Matrix Multiplication

#### Neural Network Example:

A single layer neural network:
$$
\text{output} = \text{activation}(X \cdot W + b)
$$

* X = input matrix (sample * features)
* W = weights matrix (features x neurons)
* b = bias vector
* . = matrix multiplication

instead of computing each neuron individually, matrix multiplication lets us compute all neurons for all samples at once --super efficient

````python
# Example Neural net layers

# 3 samples , 2 features
X  = np.array([[1, 2],
              [3, 4],
              [5, 6]])

#2 neurons in the layer
W = np.array([[0.1, 0.2], 
              [0.3, 0.4]])

b = np.array([0.5, 0.6])

# Linear combination
Z = np.dot(X, W) + b 
Z
````

## Note:
* Rows of first x columns of second --> new matrix
* Dimensions must match (column of first = rows of second)
* Matrix multiplication is everywhere in ML:
     * Fully connected layers
     * Convolution layers (as im2col tricks)
     * Attention mechanism in transformers
