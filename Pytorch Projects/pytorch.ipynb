## Autograd

```python
x = torch.randn(3)
print(x)
```

```python
x = torch.randn(3, requires_grad=True)
print(x)
```

```python
y = x+2
print(y)

z = y*y*2
print(z)
z = z.mean()
print(z)
```

```python
z.backward() #dz/dx
```

```python
print(x.grad)
```
