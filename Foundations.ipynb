## Algebra Basics

### Variable

* A variable is just a symbol of container that stores values
* In Ml variables can represent features (inputs), weights or outputs

#### Example:

```python
x = 1200
y = 300000

print("House size:", x)
print("House price:", y)
```

## Equations

* An equation relates variabe together.
* In Ml, We often start with a linear equation:
                 `y = w * x + b`
  * x = input features(house price).
  * y = output (predicted price)
  * w = weight (how strongle size affects price)
  * b = bias (base price even if size = 0)

#### Example:

```python
w = 200 #weight
b = 50000 # bias
x = 1200 #house size

y_pred = w*x+b
print("Predicted house price:", y_pred)
```

## Functions

* A function is reusable equation or set of instructions.
* In ml, models are just functions that map inputs ---> outputs

```python
def predict_price(x, w, b):
    """Predict house price given size x, weight w, and bias b"""
    return w * x+b
```

```python
print(predict_price(1200, 200, 50000))
print(predict_price(1500, 200, 50000))
```

## Multiple Variables (Multivaraite equation)

* When we have many features:
    ## `y = w1x1 + w2x2 + w2x2 + b

* Example :
    * x1 = size,
    * x2 = bed rooms,
    * x3 = location_score.

```python
import numpy as np

def predict_price_multi(x, w, b):
    """x: features as array, w: weigths as array"""
    return np.dot(w, x) + b # dot products = w1*x1 + w2 * x2 + ...
    
# Example : [size, bedrooms, location_score]
x = np.array([1200, 3, 8])
w = np.array([200, 10000, 5000])
b = 30000

print("Predicted price:", predict_price_multi(x, w, b))
```

### Exponentials and logarithms

* An exponential function has the form:
          `f(x) = a^x (a>0, a!= 1)`
  * a = base (often e ~ 2.718 in ML)
  * x = exponent (controls growth/decay)

* Inverse of exponenetials: `y = loga(x) <==> a^y = x`

  * Most common base: natural log(lm, base e).
  * Logs compress large numbers.
  * Turn multiplication into addition
        `log(xy) = log(x) + log(y)`
    
      

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Explonential function: f(x) = a^x
x = np.linspace(-2, 3, 100) # x values from -2 to 3
a = 10                       # base of exponential
y_exp = a ** x               # exponential function
print(y_exp)

print("\n")
print("=========")
# 2. Logarithm: inverse of exponentital
x_log = np.linspace(0.1, 10, 100) # can't take log of 0 or negative
y_log = np.log(x_log)              # natural log (base e)
print(y_log)


# Plot exponential
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(x, y_exp, label="2^x")
plt.axhline(1, color="gray", linestyle="--")
plt.axvline(0, color="gray", linestyle="--")
plt.title("Exponential Function")
plt.xlabel("x")
plt.ylabel("2^x")
plt.legend()

# Plot logarithm
plt.subplot(1,2,2)
plt.plot(x_log, y_log, label="ln(x)")
plt.axhline(0, color="gray", linestyle="--")
plt.axvline(1, color="gray", linestyle="--")
plt.title("Logarithm Function (ln)")
plt.xlabel("x")
plt.ylabel("ln(x)")
plt.legend()

plt.show()
```

```python
import numpy as np

print("ln(1) =", np.log(1))   # 0.0
print("ln(e) =", np.log(np.e)) # 1.0

# what happens near 0?
for val in [0.1, 0.01, 0.001]:
    print(f"ln({val}) = {np.log(val)}")

# # try negative input (will give nan with warning)
print("ln(-1) =", np.log(-1))
```

## Functions and Transformations

### Linear Functions

   `f(x) = ax + b`

* a = slope (rate of change)
* b = intercept (starting value)

Graph: Straight line
'ML Applications'
* Linear regression --> predict house pirces:

      price = a.size + b


* Neuaral network layers --> each layers is often a linear transform before activation.

       y = Wx + b


### Quadratic Functions
  
    f(x) = ax^2 + bx + c
____________
* Graph: parabola (U-shape or upside-down)
* Has a minimum/maximum.

> Important property: Captures curvature.

'ML Applications'
* Loss functions --> Mean squared error (MSE):
* Support vector machines (SVMs) use quadratic optimization


### Exponential Functions

      f(x) = a^x (a>0. a!=1)
* Grows(or decays) very fast.
* Base e ~ 2.718 is common in ML

'ML Applications'
Learning rate decay:
       
        ηt​=η0​⋅e^−λt



(learning rate shrinks exponentially with time)

Softmax function:

	​softmax(zi​)=​e^zi​​/∑j​e^zj

Converts scores into probabilities.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 200)
print(x)
print("-----------")

# Functions
linear = 2*x + 1
print(linear)
quadratic = x**2 - 2*x + 1
print(quadratic)
exponential = np.exp(x)
print(exponential)

plt.figure(figsize=(10, 6))

plt.plot(x, linear, label="Linear: 2x-1")
plt.plot(x, quadratic, label="Quadratic: x^2-2X+1")
plt.plot(x, exponential, label="Exponential: e^x")


plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.legend()
plt.title("Linear vs Quadratic vs Exponential Functions")
plt.show()
```

Linear → straight-line relationships (regression, NN layers).

Quadratic → curvature, optimization, loss functions.

Exponential → growth/decay, softmax, learning rate schedules.

## Coordinate Geometry in ML

Coordinate geometry = combining algebra + geometry to analyze positions of points, lines, and shapes in a plane.
in ML, this is useful for distances, angles, similarity measures.

### 1.Slope of a line
slope tells us how steep a line is.

* Formula for slope between two points
 
        `m = y2-y1/x2-x1`

* Example: PointsA(1,2), B(4,5) 
 
        `m = 5-2/4-1 = 3/3 = 1`

ML Applications: Slope is used in linear regression (the coefficient w in y = wx + b)

### 2. Distance Between Two Points
Euclidean distance formula:
              
        `d = \/(x2 - x1)^2 + (y2-y1)^2`

* Example: Points A(1,2), B(4,6):

        `d = \/(4-1)^2 + (6-2)^2 = \/9+16 = 5`

ML Applications: Distance is the basis of KNN (k-nearest neighbors), clustering, and similarity search

### 3. Dot Product of Vectors
if a = (x1, y1) and b = (x2, y2):
            
            `a.b = x1x2 + y1y2`
Also:
             `a.b = |a||b|cosθ
            
* Example: a = (1,2), b = (3,4)
                a.b = 1.3+2.4 = 11
    
ML Connection:
* Cosine similarity (used in NLP, Recommendation systems):

                    `cosθ = a.b/|a||b|

```python
import numpy as np

# Points
A = np.array([1,2])
B = np.array([4,6])

# 1. Slope
slope = (B[1] - A[1]) / (B[0] - A[0])
print("Slope", slope)

# 2. Distance
distance = np.linalg.norm(B - A) # Euclidean norm
print("Distance: ", distance)

# 3. Dot product
a = np.array([1,2])
b = np.array((3,4))
dot = np.dot(a, b) 
cosine_sim = dot /(np.linalg.norm(a) * np.linalg.norm(b))

print("Dot Product:", dot)
print("Cosine Similarity:", cosine_sim)
```

    ` x1+x2=10 and 2x1−x2=3 `

```python
import numpy as np

A = np.array([[1,1], [2, -1]])

B = np.array([10, 3])

sol = np.linalg.solve(A, B)
sol
```

```python
# above code without using linalg in numpy

#coefficient matrix
A = [[1,1],
     [2, -1]]

# Constant vector
B = [10, 3]

# Calculate determinant of 2 x 2 matrix A
det = A[0][0]*A[1][1] - A[0][1]*A[1][0]

if det == 0:
    print("No unique solution exists")
else:
    # Calculate inverse of A manually for 2x2
    A_inv = [[A[1][1]/det, -A[0][1]/det],
             [-A[1][0]/det, A[0][0]/det]]
    
    # Multiply A_inv by B to get solution vector
    x1 = A_inv[0][0]*B[0] + A_inv[1][1]*B[1]
    x2 = A_inv[1][0]*B[0] + A_inv[0][0]*B[1]

    print(x1, x2)
```
